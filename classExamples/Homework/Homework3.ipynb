```python
import pandas as pd

dfTitles = pd.read_csv('http://128.198.162.191/dataFiles/IMDB/titleBasics.tsv', sep='\t', low_memory=False)
dfRatings = pd.read_csv('http://128.198.162.191/dataFiles/IMDB/ratings.tsv', sep='\t', low_memory=False)
dfNames = pd.read_csv('http://128.198.162.191/dataFiles/IMDB/nameBasics.tsv', sep='\t', low_memory=False)
dfCrew = pd.read_csv('http://128.198.162.191/dataFiles/IMDB/crew.tsv', sep='\t', low_memory=False)
```

## Question 1: Movie Writer with Highest Number of Votes and Highest Rating Find the movie Writer with the highest number of votes and the highest rating.

```python
dfRatings.head() 
dfCrew.head()
```

### Merging Ratings and Crew

```python
ratingsAndCrew = pd.merge(dfRatings, dfCrew, on='tconst')

ratingsAndCrew.head()
```

### Merge with Titles for titleType

```python
ratingsCrewTitles = pd.merge(ratingsAndCrew, dfTitles[['tconst', 'titleType']], on='tconst')

movieWritersOnly = ratingsCrewTitles.query('titleType == "movie"')
```

### Renaming Writers column

```python
dfWriterRename = movieWritersOnly.rename(columns={'writers': 'nconst'})
```

### Merging writer Names

```python
ratingsCrewWriter = pd.merge(dfWriterRename, dfNames, on='nconst') 
sortVotes = ratingsCrewWriter.sort_values(by=['numVotes', 'averageRating'], ascending=False)
sortVotes.head()
```

### Sorting and filtering

```python
# Sort by highest votes then highest rating 
semi1 = ratingsCrewWriter.sort_values(by=['numVotes', 'averageRating'], ascending=False) 

# Select required columns in the requested order: primaryName, numVotes, averageRating 
semi2 = semi1[['primaryName', 'numVotes', 'averageRating']] 

# Select top result 
final_q1 = semi2.head(1) 
final_q1
```

### Exporting to file

```python
final_q1.to_csv('question1.csv', index=False)
```

## Question 2: Highest-Rated Film for 2000 (>= 50,000 Votes) Find the highest-rated movie released in the year 2000 with at least 50,000 votes, sorted by rating and votes.

```python
dfTitles.head(1) 
dfRatings.head(1)
```

### Merge Titles and Ratings

```python
highRated = pd.merge(dfTitles, dfRatings, on='tconst') 

highRated.head()
```

### Filter Movies in 2000 with at least 50,000 votes

```python
movieFilter = highRated.query('x0')

movieFilter.head()
```

### Sort and format output

```python
# Sort by rating then votes 
semi1 = movieFilter.sort_values(by=['averageRating', 'numVotes'], ascending=False) 

# Subset columns (startYear, primaryTitle, numVotes, averageRating) 
semi2 = semi1[['startYear', 'primaryTitle', 'numVotes', 'averageRating']] 

# Extract the top film (use .head(1) for the single highest, or .head(5) if viewing the top 5) 
final_q2 = semi2.head(1) 
final_q2
```

### Export results

```python
final_q2.to_csv('question2.csv', index=False)
```

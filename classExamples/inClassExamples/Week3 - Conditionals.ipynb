# Conditionals

## IF Statements
IF statements check our conditional expression and only execute the code statements on a TRUE.

```python
userName = input("What is your name?")
userAge = int(input('What year were you born?'))

age = 2026 - userAge

print("Hello " + userName)
print("You are " +str(age)+ " years old.")

print("This is the last statement.")

# IF statement example
if age >= 21:
    print ("Welcome to the bar, enjoy your cold snack.")
```

## Two-Way IF statements
Checks the conditions and executes statement depending on TRUE or FALSE.

```python
userName = input("What is your name?")
userAge = int(input('What year were you born?'))

age = 2026 - userAge

print("Hello " + userName)
print("You are " +str(age)+ " years old.")


# Two-Way IF Statement example
if age >= 21:
    print("Welcome, enjoy the cold snacks!")
else:
    print("Sorry, no cold snack for you.")
```

## Ternerary expressions
Shorter versions (slightly) of the same function.

```python
userName = input("What is your name?")
userAge = int(input('What year were you born?'))

age = 2026 - userAge

print("Hello " + userName)
print("You are " +str(age)+ " years old.")

# Ternary expression example
canEnter = "You can enter the bar" if age >= 21 else "You cannot enter."

print(canEnter)
```

## Nested IF statements


```python
userName = input("What is your name?")
userAge = int(input('What year were you born?'))

age = 2026 - userAge

print("Hello " + userName)
print("You are " +str(age)+ " years old.")

# Nested IF example
if age >= 18:
    if age <= 25:
        print("You can enlist in the service.")
    else:
        print("You are too old to enlist.")
else:
    print ("You are not eligible to enlist.")
```

## Else-IF statements

```python
userName = input("What is your name?")
userAge = int(input('What year were you born?'))

age = 2026 - userAge

print("Hello " + userName)
print("You are " +str(age)+ " years old.")

# Else-IF example
if age <= 16:
    print("You are too young to drive.")
elif age >= 16 and age <= 18:
    print ("You are eligible for a learners permit.")
# Add an additional requirement
elif age >25 and age <= 70:
    print("You may be eligible for an insurance discount.")
else:
    print("You are eligible for a drivers license.")
```

## Case switch (match) statements

```python
dayToView = int(input('What number (0 to 6) would you like to show?'))

day = dayToView

day_of_week = ""

if day == 0:
    day_of_week = 'Sunday'
elif day == 1:
    day_of_week = 'Monday'
elif day == 2:
    day_of_week = 'Tuesday'
elif day == 3:
    day_of_week = 'Wednesday'
elif day == 4:
    day_of_week = 'Thursday'
elif day == 5:
    day_of_week = 'Friday'
elif day == 6:
    day_of_week = 'Saturday'
else:
    day_of_week = "Invalid day selection"

print(day_of_week)
```

Shorthand version of case statements

```python
dayToView = int(input('What number (0 to 6) would you like to show?'))

day = dayToView

day_of_week = ""

case = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}

day_of_week = case.get(day, 'Invalid Day')

print(day_of_week)
```

# ATM Simulator

```python
# Initialize an account balance
balance = 1000

# While loop to keep the menu visible.
while True:
    print("Welcome to the ATM simulator.")
    print("1. Check Balance")
    print("2. Withdraw")
    print("3. Deposit")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

# Conditionals to act appropriately based on user selection.
    if choice == 1:
        print("Your current balance $" + str(balance))
    elif choice == 2:
        amount = int(input("Enter the amount to withdraw: "))
        if amount > balance:
            print("Insufficient funds!")
        elif amount <= 0:
            print("Invalid amount")
        else:
            balance -= amount
            print("$" + str(amount) + " withdrawn successfully.")
            print("Your remaining balance is $" + str(balance))
    elif choice == 3:
        amount = int(input("Enter the amount to deposit: "))
        if amount <= 0:
            print("Invalid amount")
        else:
            balance += amount
            print("$" + str(amount) + " deposited successfully.")
            print("Your remaining balance is $" + str(balance))
    elif choice == 4:
        print("Thank you for banking with us.")
        break
    else:
        print("Invalid choice. Please try again.")

        # Seomthing
```

# Who survived the Titanic sinking

```python
# Grouping by survivors and NOT
who_survived = df.groupby('Survived')['Name'].apply(list)

# Print out the names
print('List of Survivors: ', who_survived[1])
print('List of those not experiencing life: ', who_survived[0])
```

```python
# Create the lists with one passenger name on each row.

print("Passengers who surived:")
for name in who_survived[1]:
    print(name)

print('\nList of passengers that did not survive: ')
for name in who_survived[0]:
    print(name)
```

## Odds of survival by sex

```python
df.head(2)
```

```python
survivedBySex = df.groupby(['Survived', 'Sex']).size()

print(survivedBySex)
```

### Odds by sex

```python
# We need to filter the data for survivors and casualities
maleSurvived = df[(df['Sex'] == 'male') & (df['Survived'] == 1)].shape[0]
maleDied = df[(df['Sex'] == 'male') & (df['Survived'] == 0)].shape[0]


# Female survivors
femaleSurvived = df[(df['Sex'] == 'female') & (df['Survived'] == 1)].shape[0]
femaleDied = df[(df['Sex'] == 'female') & (df['Survived'] == 0)].shape[0]

# Calculate the odds of survival
maleOdds = maleSurvived / maleDied
femaleOdds = femaleSurvived / femaleDied



# Print out the results
print(f'Odds of survival for Men:  {maleOdds:.2f}')
print(f'Odd of survival for Women: {femaleOdds:.2f}')
```

## What was the name and year of the movie with the hightest rating and highest number of votes.
To accomplish we're going to have to use data from multiple files.

```python
# Merge two dataFrames together

mvRating = pd.merge(dfTitles, dfRatings, on='tconst')

mvRating
```

```python
# Reduce the dataFrame to only movies
ratingSort = (mvRating.query('titleType == "movie"'))

ratingSort
```

```python
# Sort the dataFrame on ratings.

ratingSort.sort_values(by='averageRating', ascending=False)
```

```python
# Sort by rating and votes
ratingSort.sort_values(by=['averageRating', 'numVotes'], ascending = False)
```

```python
ratingSort.sort_values(by=['numVotes', 'averageRating'], ascending = False)
```

## Prep data for export

```python
# Create a dataFrame with the queried and sorted data

semi1 = (ratingSort.sort_values(by=['numVotes', 'averageRating'], ascending=False))


# Create a dataFrame that only has the required columns (labels)
semi2 = semi1[['primaryTitle', 'startYear', 'averageRating', 'numVotes']]

# Select the subset of the data
final = (semi2.head())

print(final)
```

```python
# Output the results to a file

final.to_csv('files/finalResults.csv', index=False)
```

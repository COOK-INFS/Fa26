# Strings

Strings can behave a little differently that other data types.


## Basic String

```python
myString = "Hello World"
print(myString)
```

## Retrieving a specific index location for a String

```python
myString = "Hello World"

print(myString[3])
```

## Retrieve a "slice" of a String

```python
myString = "Hello World"

print(myString[0:5])
```

## Strings are immutable

```python
# Initial strings

string1 = "Bob"
string2 = "Mary"

# Memory locations before change
memory_before_string1 = id(string1)
memory_before_string2 = id(string2)

# Changing string1 to "Mary"
string1 = "Mary"

# Memory locations after change
memory_after_string1 = id(string1)
memory_after_string2 = id(string2)

# Output the memory locations
print(f"Memory location of 'string1' before change: {memory_before_string1}")
print(f"Memory location of 'string2' before change: {memory_before_string2}")
print(f"Memory location of 'string1' after change: {memory_after_string1}")
print(f"Memory location of 'string2' after change (should be unchanged): {memory_after_string2}")
```

## Formatting Strings

### Basic Concatenating of Strings

```python
name = "Alice"
age = 30
greeting = "Hello, " + name + ". You are " + str(age) + " years old."
print(greeting)
```

```python
### Formatting with the "f-string" method
```

```python
name = "Alice"
age = 30
greeting = f"Hello, {name}. You are {age} years old."
print(greeting)
```

### Formatting with the ".format" method

```python
name = "Alice"
age = 30
greeting = "Hello, {}. You are {} years old.".format(name, age)
print(greeting)
```

# Escaping characters


You can use the ability to "escape" characters that are seen by the Python interpreter as special characters and therefore cause issues in strings of text.

## Failing to escape characters

```python
escapeEx = "I am a "goof", no doubt about it."
print(escapeEx)
```

## Corrected escaping characters

```python
escapeEx = "I am a \"goof\", no doubt about it."
print(escapeEx)
```

# Newline and Carriage return

## Newline

```python
escapeEx = "This line is line 1. \nThis line is line 2."
print(escapeEx)
```

## Carriage return

```python
escapeEx = "This line is line 1. \rThis line is line 2."
print(escapeEx)
```

# Searching in Strings

## Searching using the ".find" method

```python
my_string = "Hello, World!"
index = my_string.find("World")
print(index)
```

## Searching using the "in" method

```python
my_string = "Hello, World!"
result = "World" in my_string
print(result)
```

# Files examples

## Writing Files

This example creates a file and writes data to it. It is important to note that if no path is specified the file will be created in the root of the directory the program is in.

```python
f = open("files/readFileExample.txt", "w")
f.write("My first line of text.")
f.close()
```

## Reading Files

```python
f = open("files/readFileExample.txt", "r")
print(f.read())
```

```python
# Another reading example using the "with" method

text = "Here I go again"

with open('files/readFileExample.txt', 'w') as f:
    f.write(text)
with open('files/readFileExample.txt', 'r') as f:
    print(f.read())
```

## Appending data to a file using the "with" method

```python
text = "And more text"

with open('files/readFileExample.txt', 'a') as f:
    f.write(text)
with open('files/readFileExample.txt', 'r') as f:
    print(f.read())
```

## Error Handling

When we are dealing with things like reading files, it is a good idea to make our errors more understandable to our users.

```python
try:
    with open('files/readFileExample.txt','r') as f:
        print(f.read())
except FileNotFoundError:
    print("files/That file was not found!")
```

# Config file example

```python
# Search parameter
searchItem = 'theme'

# Open the file in read mode
with open('files/myConfig.ini', 'r') as f:
    lines = f.readlines() # Read all lines in the file

# Search for the 'theme' setting
for line in lines:
    if searchItem in line:
        _, searchItem = line.split("=", 1) # Ignore the values left of the "=" and assign to "_"
        searchItem = searchItem.strip()
        break

# Display results or lack of results
if searchItem:
    print(f"The theme setting is: {searchItem}")
else:
    print("The theme setting was not found.")
```

# Config parser example

```python
# First ensure you have installed the configparser
# uv add configparser


from configparser import ConfigParser

section='Settings'
configItem='theme'

# Initialize the parser
config = ConfigParser()

# Read the config file
config.read('files/myConfig.ini')

# Retrieve the setting
theme = config.get(section, configItem)
print(f"The setting is: {theme}")
```

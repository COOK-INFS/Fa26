### Arithmetic operators

````python
# Assigning values to my variables
x = 5
y = 3


# Add the the two variables
print(x+y)

# Subtract the variables
print(x-y)

# Multiply the variables
print(x*y)

# Divide the variables
print(x/y)

# Return x to the power of y
print(x ** y)

# Return the "remainder of x/y"
print (x % y)
````

### Scientific Notation

````python
x = 10

x **= 3  # Creates a value of 'x' to the power of 3.

print(x)
````

### Divisible by

````python
x = 10

x //= 3 # How many times will 3 "go into" x.

print(x)
````

````python
x = 10

x %= 3 # Shows the remainder of 3 into 'x'.

print(x)
````

## Comparison operators

````python
x = 10
y = 3

# Is x equal to y?
print(x == y)

# x IS NOT EQUAL to y.
print (x != y)

# Greater than
print (x > y)

# Less than
print (x < y)

# Greater than or equal to
print (x >= y)
````

## Logical Operators

````python
# AND operator

x = 10

# AND operations require that ALL conditions must be TRUE.
print (x > 2 and x < 11)
````

````python
# Logical OR operator

x = 10

# In an OR, any of the conditions can be true to receive a TRUE output.
print (x > 2 or x < 9)
````

````python
# NOT logical operator

x = 10

# This will return the opposite of a TRUE condition

print (not(x > 2 and x < 11))
````
